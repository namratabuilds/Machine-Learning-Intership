```python
import pandas as pd

df = pd.read_csv("Book1.csv")

# Display only Name and Marks
result = df[["Student", "Math"]]

print(result)
```

```python
#Fill Missing
df.fillna("Unknown")
```

```python
#Check Missing Values
print(df.isnull())
```

```python
#Count Missing
print(df.isnull().sum())
```

```python
#Marks greater than 80
df[df["Marks"]>80]

#Multiple Conditions
df[(df["Marks"]>80) & (df["Roll"]>18)]
```

```python
#Ascending
df.sort_values("Marks")
#Descending
df.sort_values("Marks",ascending=False)
```

```python
import pandas as  pd 
df=pd.read_csv("employee.csv")
print(df)
```

```python
df["Bonus"]=5000
df["Total"]=df["salary"]+df["Bonus"]
print(df)
```

```python
#Average Salary
df.groupby("department")["salary"].mean()
```

```python
#Sum
df.groupby("department")["salary"].sum()
```

```python
#Maximum
df.groupby("department")["salary"].max()
```

```python
#Count
df.groupby("department")["salary"].count()
```
