```python
#Student Marks Prediction
from sklearn.linear_model import LinearRegression

hours = [[1],[2],[3],[4],[5]]

marks = [20,35,50,65,80]

model = LinearRegression()

model.fit(hours, marks)

prediction = model.predict([[6]])

print("Predicted Marks =", prediction[0])
```

```python
#Model Parameters
print(model.coef_)

print(model.intercept_)
```

```python
#Accuracy
print(model.score(hours,marks))
```

```python
from sklearn.linear_model import LogisticRegression

X = [[2],[3],[4],[5],[6],[7]]

y = [0,0,0,1,1,1]

model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[5]])

print(prediction)
```

```python
#Predict Probability
print(model.predict_proba([[5]]))
```

```python
from sklearn.linear_model import LogisticRegression

X = [[2], [3], [4], [5], [6], [7]]
y = [0, 0, 0, 1, 1, 1]

model = LogisticRegression()

model.fit(X, y)

#predict() vs predict_proba()
prediction = model.predict([[5]])
probability = model.predict_proba([[5]])

print("Prediction:", prediction)
print("Probability:", probability)
```

```python
#Predicting Spam Emails 
from sklearn.linear_model import LogisticRegression

# Number of Spam Words
X = [[2],[4],[6],[8],[10],[12]]

# 0 = Not Spam, 1 = Spam
y = [0,0,0,1,1,1]


model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[9]])

print("Spam Prediction =", prediction[0])

print("Probability =", model.predict_proba([[9]]))
```

```python
#Predicting Customer Churn
from sklearn.linear_model import LogisticRegression

# Months Since Last Purchase
X = [[1],[2],[3],[8],[10],[12]]

# 0 = Stay, 1 = Leave
y = [0,0,0,1,1,1]

model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[9]])

print("Customer Churn Prediction =", prediction[0])

print("Probability =", model.predict_proba([[9]]))
```
