```python
#Accuracy
```

```python
from sklearn.metrics import accuracy_score

actual = [1,1,0,0,1]

predicted = [1,0,0,0,1]

accuracy = accuracy_score(actual, predicted)

print("Accuracy =", accuracy)
```

```python
#Student Result Prediction 
from sklearn.metrics import accuracy_score

# Actual Results
actual = [
    "Pass",
    "Pass",
    "Fail",
    "Fail",
    "Pass"
]

# Predicted Results
predicted = [
    "Pass",
    "Fail",
    "Fail",
    "Fail",
    "Pass"
]

accuracy = accuracy_score(actual, predicted)

print("Accuracy =", accuracy)
```

```python
#Weather Prediction 
from sklearn.metrics import accuracy_score

# Actual Weather
actual = [
    "Rainy",
    "Rainy",
    "Sunny",
    "Sunny",
    "Rainy"
]

# Predicted Weather
predicted = [
    "Rainy",
    "Sunny",
    "Sunny",
    "Sunny",
    "Rainy"
]

accuracy = accuracy_score(actual, predicted)

print("Accuracy =", accuracy)
```

```python
#Precision
```

```python
from sklearn.metrics import precision_score

actual = [1,1,0,0,1]

predicted = [1,0,0,0,1]

print(precision_score(actual,predicted))
```

```python
#Recall
from sklearn.metrics import recall_score

print(recall_score(actual,predicted))
```

```python
#F1 Score
from sklearn.metrics import f1_score

print(f1_score(actual,predicted))
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

# Actual Result
actual = [
    1,
    1,
    0,
    0,
    1
]

# Predicted Result
predicted = [
    1,
    0,
    0,
    0,
    1
]

accuracy = accuracy_score(actual, predicted)
precision = precision_score(actual, predicted)
recall = recall_score(actual, predicted)
f1 = f1_score(actual, predicted)

print("Accuracy =", accuracy)
print("Precision =", precision)
print("Recall =", recall)
print("F1 Score =", f1)
```

```python
#Feature Scaling
```

```python
#Min-Max Scaling
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

data = pd.DataFrame({
    "Marks":[40,60,80,100]
})

scaler = MinMaxScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

```python
#House Price Prediction
from sklearn.preprocessing import MinMaxScaler

# House Size (sq ft)
X = [
    [800],
    [1200],
    [1500],
    [2000],
    [2500]
]

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(X)

print(scaled_data)
```

```python
#Standardization (Z-Score Scaling)
from sklearn.preprocessing import StandardScaler
import pandas as pd

data = pd.DataFrame({
    "Marks":[40,60,80,100]
})

scaler = StandardScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

```python
#Salary Prediction
from sklearn.preprocessing import StandardScaler

# Salary (₹)
X = [
    [25000],
    [35000],
    [45000],
    [55000],
    [65000]
]

scaler = StandardScaler()

scaled_data = scaler.fit_transform(X)

print(scaled_data)
```
