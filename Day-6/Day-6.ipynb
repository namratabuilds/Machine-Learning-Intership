```python
#HOUSE PRICE PREDICTION BASED ON HOUSE SIZE
from sklearn.linear_model import LinearRegression

# House size
X = [[1000], [1200], [1500], [1800]]

# House price (in lakh)
y = [40, 50, 65, 80]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[1600]])

print("Predicted Price:", prediction[0], "Lakh")
```

```python
#K-Means Clustering
```

```python
from sklearn.cluster import KMeans
import numpy as np

data = np.array([
    [2, 3],
    [3, 4],
    [10, 12],
    [11, 13]
])

model = KMeans(n_clusters=2, random_state=42)
model.fit(data)

print("Cluster Labels:")
print(model.labels_)
```

```python
#Customer Shopping Clustering 
from sklearn.cluster import KMeans
import numpy as np

# Age, Monthly Spending (₹)
data = np.array([
    [20, 5000],
    [22, 6000],
    [45, 30000],
    [50, 35000]
])

model = KMeans(n_clusters=2, random_state=42)
model.fit(data)

print("Cluster Labels:")
print(model.labels_)
```
