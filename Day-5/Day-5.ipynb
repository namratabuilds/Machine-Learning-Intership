```python
#Adding Labels

import matplotlib.pyplot as plt
students = ["Amit","Rahul","Priya","Neha"]
marks = [75,82,91,68]
plt.plot(students, marks)
plt.title("Student Marks")

#Change Line Color
plt.plot(students, marks, color="red")

#Change Line Width
plt.plot(students, marks, linewidth=5)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Change Line Style

import matplotlib.pyplot as plt
students = ["Amit","Rahul","Priya","Neha"]
marks = [75,82,91,68]
plt.plot(students, marks)
plt.plot(students, marks, linestyle="--")

#Add Markers
plt.plot(students, marks, marker="o")
plt.show()
```

```python
#Combine Everything
import matplotlib.pyplot as plt
students = ["Amit","Rahul","Priya","Neha"]
marks = [75,82,91,68]
plt.plot(
students,
marks,
color="blue",
linewidth=3,
linestyle="--",
marker="o"
)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Pie Chart

import matplotlib.pyplot as plt
subjects = ["Math","Science","English"]
marks = [40,35,25]
plt.pie(marks, labels=subjects)
plt.show()
```

```python
#Multiple Lines

import matplotlib.pyplot as plt
months = [1,2,3,4,5]
sales1 = [10,20,25,35,40]
sales2 = [15,18,30,32,45]
plt.plot(months, sales1)
plt.plot(months, sales2)
plt.show()
```

```python
#1. Draw a line chart for the temperatures of seven days
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temp = [30, 32, 31, 29, 33, 35, 34]

plt.plot(days, temp)
plt.title("Temperature of Seven Days")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
#2. Create a bar chart showing the number of books read by five students.
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D", "E"]
books = [5, 8, 4, 7, 6]

plt.bar(students, books, color="skyblue")
plt.title("Books Read by Students")
plt.xlabel("Students")
plt.ylabel("Number of Books")
plt.show()
```

```python
#3. Draw a pie chart of your monthly expenses.
import matplotlib.pyplot as plt

categories = ["Food", "Rent", "Transport", "Shopping", "Others"]
expenses = [5000, 12000, 3000, 4000, 2000]

plt.pie(expenses, labels=categories, autopct="%1.1f%%")
plt.title("Monthly Expenses")
plt.show()
```

```python
#4. Plot a scatter graph for study hours versus marks.
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 45, 50, 60, 70, 85, 95]

plt.scatter(hours, marks, color="red")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
#5. Create a histogram using the marks of 20 students.
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75,
         78, 80, 82, 85, 87, 90, 91, 93, 95, 98]

plt.hist(marks, bins=5, color="orange", edgecolor="black")
plt.title("Histogram of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
#8. Display two lines on the same graph with a legend.
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")
plt.title("Two Lines on Same Graph")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.rand(10,10)
plt.imshow(x,cmap="Set3",interpolation="nearest",alpha=0.5)
plt.title("HeatMap using imshow() Function")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.rand(10,10)
plt.imshow(x,cmap="Set3",interpolation="bicubic",alpha=0.5)
plt.title("HeatMap using imshow() Function")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.random.rand(10,10)
plt.pcolormesh(x,cmap="hsv",alpha=0.8,edgecolor="k")
plt.title("HeatMap using pcolormesh() Function")
plt.show()
```
